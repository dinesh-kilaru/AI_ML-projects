```python
def cap_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[column] = np.where(df[column] < lower_bound, lower_bound, df[column])
    df[column] = np.where(df[column] > upper_bound, upper_bound, df[column])
    return df

df_treated = df.copy()
df_treated['Age'] = df_treated['Age'].fillna(df_treated['Age'].median())
df_treated['Fare'] = df_treated['Fare'].fillna(df_treated['Fare'].median())

df_treated = cap_outliers_iqr(df_treated, 'Age')
df_treated = cap_outliers_iqr(df_treated, 'Fare')

display(df_treated.describe())
```
